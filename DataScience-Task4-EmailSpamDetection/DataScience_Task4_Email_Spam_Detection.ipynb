### 3. Feature Extraction (TF-IDF) & Mathematical Formulation
TF-IDF (Term Frequency - Inverse Document Frequency) penalizes words that appear ubiquitously across all documents while emphasizing terms that carry discriminative semantic weight:
$$\text{TF-IDF}(t, d, D) = \text{TF}(t, d) \times \log\left(\frac{1 + |D|}{1 + |\{d \in D : t \in d\}|}\right) + 1$$

### 5. Critical Technical Discussion: The Precision-Recall Dilemma in Spam Detection
> **Why is Precision often prioritized alongside Recall in Spam Detection?**
> - In an enterprise email filtering system, a **False Positive** (classifying an important business email as spam) causes critical communication loss, missed contractual deadlines, or lost customer deals.
> - A **False Negative** (allowing an occasional spam email into the inbox) causes minor annoyance.
> - Therefore, optimal spam filters require **$\ge 99\%$ Precision** while maintaining robust Recall.
